# BAF dataset profiling

This notebook profiles Base and Variants I-V, records dtypes and `-1` missing-code counts, and compares every variant with Base. The reusable implementation is in `ai4cyber.profiling`.

```python
import json
from pathlib import Path

import pandas as pd

from ai4cyber.config import TrainingConfig
from ai4cyber.profiling import run_profiling
```

```python
root = Path.cwd()
if not (root / 'configs').exists():
    root = root.parent
config = TrainingConfig.from_json(root / 'configs/training.json')
manifest = run_profiling(
    root / 'data/raw', config, root / 'outputs/profiling'
)
manifest['source']
```

```python
pd.read_csv(root / 'outputs/profiling/dataset_comparison.csv')
```

```python
base = json.loads((root / 'outputs/profiling/base_profile.json').read_text())
pd.DataFrame({
    'dtype': pd.Series(base['dtypes']),
    'minus_one_count': pd.Series(base['minus_one_counts_all_numeric']),
    'documented_sentinel_count': pd.Series(base['sentinel_minus_one_counts']),
}).fillna(0)
```
